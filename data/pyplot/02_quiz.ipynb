```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False
```

문제 1. 1월부터 6월까지의 월별 지출 50, 70, 60, 90, 80, 100을 선 그래프로 그리고, 제목과 x축, y축 라벨을 붙이세요.

```python
df = pd.DataFrame({
    "달": ["1월", "2월", "3월", "4월", "5월", "6월"],
    "지출": [50, 70, 60, 90, 80, 100]
})

df.plot(x="달", y="지출", kind="line", title="월별 지출")
```

문제 2. 과일별 판매량 {"사과": 30, "바나나": 45, "포도": 20, "귤": 55}을 막대 그래프로 그리세요.

```python
data = {
    "사과": 30,
    "바나나": 45,
    "포도": 20,
    "귤": 55
}

plt.plot(data.keys(), data.values())
plt.xlabel("과일")
plt.ylabel("판매량")
plt.show()
```

문제 3. 학생 20명의 시험 점수 리스트를 임의로 만들고(직접 값을 넣어도 됨), 히스토그램으로 점수 분포를 그리세요. (구간은 5개로)

```python
import numpy as np

scores = np.random.randint(0, 100, 20)
print(scores)

plt.hist(scores, bins = 5)
plt.show()
```

문제 4. 광고비 10, 20, 30, 40, 50와 매출 100, 180, 250, 300, 400의 관계를 산점도로 그리세요.

```python
ads = [10, 20, 30, 40, 50]
revs = [100, 180, 250, 300, 400]

plt.scatter(x=ads, y=revs)
plt.xlabel="광고비"
plt.ylabel="매출"
plt.show()
```

문제 5. pandas DataFrame으로 부서별 인원수({"개발": 12, "영업": 8, "디자인": 5, "인사": 3})를 만들고, df.plot(kind="pie")를 이용해 원 그래프로 그리세요.

```python
df = pd.DataFrame({
    "부서": ["개발", "영업", "디자인", "인사"],
    "인원수": [12, 8, 5, 3]
})
df.set_index("부서", inplace=True)
df.plot(df, kind="pie", subplots=True, autopct="%1.1f%%")
```
