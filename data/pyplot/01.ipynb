```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False
```

```python
months = [1, 2, 3, 4, 5, 6]
sales = [100, 150, 130, 200, 180, 250]

plt.plot(months, sales)
plt.title("월별 매출")
plt.xlabel("월")
plt.ylabel("매출(만원)")
plt.show()
```

```python
months = [1, 2, 3, 4, 5, 6 ]
store_a = [100, 150, 130, 200, 180, 250]
store_b = [80, 120, 160, 140, 190, 210]

plt.plot(months, store_a, label="A매장", marker="o")
plt.plot(months, store_b, label="B매장", marker="s", color="red")

plt.title("매장별 월 매출 비교")
plt.xlabel("월")
plt.ylabel("매출(만원)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
subjects = ["국어", "영어", "수학", "과학"]
scores = [85, 78, 92, 88]

plt.bar(subjects, scores, color="skyblue")
plt.title("과목별 점수")
plt.xlabel("과목")
plt.ylabel("점수")
plt.show()
```

```python
names = ["철수", "영희", "민수", "지훈"]
scores = [85, 92, 78, 88]

plt.barh(names, scores, color="orange")
plt.title("학생별 점수")
plt.xlabel("점수")
plt.show()
```

```python
scores = [55, 60, 62, 68, 70, 72, 75, 78, 80, 82, 85, 88, 90, 92, 95, 98]

plt.hist(scores, bins=5, color="green", edgecolor="black")
plt.title("점수 분포")
plt.xlabel("점수 구간")
plt.ylabel("학생 수")
plt.show()
```

```python
study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
scores = [50, 55, 60, 65, 70, 80, 85, 95]

plt.scatter(study_hours, scores, color="red")
plt.title("공부 시간과 점수의 관계")
plt.xlabel("공부 시간")
plt.ylabel("점수")
plt.show()
```

```python
labels = ["국어", "영어", "수학", "과학"]
values = [30, 20, 35, 15]

plt.pie(values, labels=labels, autopct="%1.1f%%", startangle=90)
plt.title("과목별 학습 시간 비율")
plt.show()
```

```python
months = [1, 2, 3, 4, 5]
sales = [100, 150, 130, 200, 180]

plt.figure(figsize=(8, 4))
plt.plot(
    months, sales,
    color="purple",
    linestyle="--",
    linewidth=2,
    marker="o",
    markersize=8
)
plt.title("월별 매출", fontsize=16)
plt.xlabel("월")
plt.ylabel("매출(만원)")
plt.ylim(0, 250)
plt.show()
```

```python
months = [1, 2, 3, 4, 5]
sales = [100, 150, 130, 200, 180]
visitors = [1000, 1200, 1100, 1500, 1400]

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot(months, sales, marker="o", color="blue")
axes[0].set_title("월별 배출")

axes[1].plot(months, visitors, marker="s", color="green")
axes[1].set_title("월별 방문자 수")

plt.tight_layout()
fig.suptitle("Monthly Report", fontsize=16, fontweight="bold", y=1.02)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Malgun Gothic"

df = pd.DataFrame({
    "월": [1, 2, 3, 4, 5],
    "매출": [100, 150, 130, 200, 180],
    "방문자": [1000, 1200, 1100, 1500, 1400],
})

df.plot(x="월", y="매출", kind="line", marker="o", title="월별 매출")
plt.show()

df.plot(x="월", y=["매출", "방문자"], kind="bar", title="월별 매출/방문자 비교")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False

df = pd.DataFrame({
    "이름": ["철수", "영희", "민수", "지훈", "수연"],
    "국어": [85, 92, 70, 88, 95],
    "영어": [78, 88, 65, 84, 90],
    "수학": [90, 95, 72, 79, 98]
})

df["평균"] = df[["국어", "영어", "수학"]].mean(axis=1)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].bar(df["이름"], df["평균"], color="skyblue")
axes[0].set_title("학생별 평균 점수")
axes[0].set_ylabel("평균 점수")

subject_avg = df[["국어", "영어", "수학"]].mean()
axes[1].pie(subject_avg, labels=subject_avg.index, autopct="%1.1f%%")
axes[1].set_title("과목별 평균 비중")

plt.tight_layout()
plt.show()

df
```
