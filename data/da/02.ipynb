```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False
```

```python
sales = pd.DataFrame({
    "매출액": [12000, 15000, 13000, 14000, 16000, 15500, 30000, 12500, 13500, 20000]
})

sales["매출액"].plot(kind="hist", bins=10, title="매출액 분포")
plt.xlabel("매출액")
plt.ylabel("빈도수")
plt.show()

sns.histplot(data=sales, x="매출액", bins=10, kde=True)
plt.title("매출액 분포 (밀도곡선 포함)")
plt.show()
```

```python
sales2 = pd.DataFrame({
    "카테고리": ["가전", "가전", "가전", "식품", "식품", "식품"],
    "매출액": [50000, 70000, 480000, 20000, 25000, 22000]
})

sales2.boxplot(column="매출액")
plt.title("전체 매출액 박스플롯")
plt.show()

sns.boxplot(data=sales2, x="카테고리", y="매출액")
plt.title("카테고리별 매출액 분포 비교")
plt.show()
```

```python
category_sales = sales2.groupby("카테고리")["매출액"].sum().reset_index()

category_sales.plot(kind="bar", x="카테고리", y="매출액", legend=False)
plt.title("카테고리별 매출액 합계")
plt.ylabel("매출액")
plt.show()

sns.barplot(data=category_sales, x="카테고리", y="매출액")
plt.title("카테고리별 매출액 합계")
plt.show()
```

```python
daily_sales = pd.DataFrame({
    "날짜": pd.date_range("2024-01-01", periods=7, freq="D"),
    "매출액": [50000, 62000, 58000, 71000, 69000, 90000, 85000]
})

daily_sales.plot(kind="line", x="날짜", y="매출액", marker="o")
plt.title("일별 매출 추세")
plt.xlabel("날짜")
plt.ylabel("매출액")
plt.show()

sns.lineplot(data=daily_sales, x="날짜", y="매출액", marker="o")
plt.title("일별 매출 추세 (seaborn)")
plt.show()
```
