```python
import pandas as pd
import numpy as np
```

```python
student_data = {
    "이름": ["철수", "영희", "민수", "지은", "우진"],
    "국어": [80, np.nan, 70, 95, 60],
    "영어": [85, 90, np.nan, 88, 75],
    "수학": [np.nan, 95, 65, np.nan, 80]
}
df = pd.DataFrame(student_data)
```

1. 결측치가 있는 위치를 True/False로 확인해 보세요.

```python
df.isnull()
```

2. 각 열(과목)별로 결측치가 몇 개씩 있는지 확인해 보세요.

```python
df.isnull().sum()
```

3. 결측치가 하나라도 있는 행을 모두 제거해 보세요. (원본은 유지)

```python
df.dropna(how="any")
```

4. "수학" 열에 결측치가 있는 행만 제거해 보세요.

```python
df.dropna(subset=["수학"])
```

5. 모든 결측치를 0으로 채워보세요.

```python
df.fillna(0)
```

6. 결측치를 "국어"는 0으로, "영어"는 해당 열의 평균으로, "수학"은 해당 열의 평균으로 각각 다르게 채워보세요.

```python
df.fillna({
    "국어": 0,
    "영어": df["영어"].mean(),
    "수학": df["수학"].mean()
})
```

7. "국어" 열의 결측치를 국어 평균값으로 채우고, 원본 df에 바로 반영해 보세요.

```python
df.fillna(value = {"국어": df["국어"].mean()}, inplace = True)
```

8. 국어, 영어, 수학 점수를 모두 더한 "총점" 열을 새로 추가해 보세요. (6번에서 결측치를 처리한 데이터 기준으로 진행)

```python
df = pd.DataFrame(student_data)
df.fillna(
    value = {
        "국어": 0,
        "영어": df["영어"].mean(),
        "수학": df["수학"].mean()
    },
    inplace = True
)
df["총점"] = df["국어"] + df["수학"] + df["영어"]
df
```

9. "총점"을 3으로 나눈 "평균" 열을 추가하고, 평균이 80 이상이면 "합격", 아니면 "불합격"이 담긴 "결과" 열도 추가해 보세요.

```python
df["평균"] = df["총점"] / 3
df["결과"] = df["평균"].apply(lambda avg: "합격" if avg >= 80 else "불합격")
df
```

10. "결과" 열을 삭제해 보세요. (원본 자체가 바로 바뀌도록)

```python
df.drop(columns = ["결과"], inplace = True)
df
```

11. "국어" 열 이름을 "한국어"로, "영어" 열 이름을 "English"로 바꿔보세요.

```python
df.rename(
    columns = {
        "국어": "한국어",
        "영어": "English"
    }, 
    inplace = True
)
df
```
