```python
import pandas as pd
import numpy as np
```

```python
fruit_data = {
    "과일": ["사과", "바나나", "포도", "딸기", "수박", "참외", "사과", "바나나"],
    "분류": ["국산", "수입", "국산", "국산", "국산", "국산", "국산", "수입"],
    "가격": [3000, 1500, 5000, 4000, 15000, np.nan, 3500, 1800], # 결측치 포함
    "재고": [10, 25, 7, 15, 3, 12, 8, 20]
}
df = pd.DataFrame(fruit_data)
```

Q1. [결측치 처리] "가격" 컬럼에 NaN(결측치) 데이터가 있습니다. 가격이 비어있는 행을 찾아서 전체 가격의 평균값으로 채워 넣어보세요. (힌트: fillna, mean 사용)

```python

mean_price = int((df["가격"].mean().round()))
df.fillna({"가격": mean_price}, inplace = True)
df["가격"] = df["가격"].astype("int64")
df
```

Q2. [정렬 및 다중 조건] "분류"가 "국산"인 과일들만 골라낸 뒤, "가격"은 비싼 순서대로(내림차순), 가격이 같다면 "재고"가 많은 순서대로(내림차순) 정렬해 보세요. (힌트: sort_values에 리스트 사용)

```python
df[df["분류"] == "국산"].sort_values(["가격", "재고"], ascending=False)
```

Q3. [파생 변수 생성] "가격"과 "재고"를 곱해서 "총재고액"이라는 새로운 컬럼을 만들어 보세요.

```python
df["총재고액"] = df["가격"] * df["재고"]
df
```

Q4. [그룹화 및 통계] "분류"(국산/수입)별로 "가격"의 평균과 "재고"의 총합합계를 각각 구해 보세요. (힌트: groupby와 agg 사용)

```python
df.groupby("분류").agg(가격평균=("가격", "mean"), 재고=("재고", "sum"))
```

```python
df.groupby("분류")["재고"].sum()
```

Q5. [조건 기반 값 변경] np.where 또는 판다스의 .loc을 활용하여, 가격이 5000원 이상이면 "프리미엄", 5000원 미만이면 "일반"이라는 값을 가지는 "등급" 컬럼을 새로 만들어 보세요.

```python
df["등급"] = df["가격"].apply(lambda price: "프리미엄" if price >= 5000 else "일반")
df
```

Q6. [텍스트 필터링] "과일" 이름에 "바"자가 포함된 과일의 행들만 필터링해 보세요. (힌트: .str.contains() 사용)

```python
df[df["과일"].str.contains("바")]
```
