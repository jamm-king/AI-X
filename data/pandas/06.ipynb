```python
data = {
    "이름": ["철수", "영희", "민수"],
    "국어": [80, 90, 70],
    "영어": [85, 95, 75]
}
df = pd.DataFrame(data)
df
```

```python
def to_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    else:
        return "C"
```

```python
df["국어등급"] = df["국어"].apply(to_grade)
df
```

```python
df["평균"] = df.apply(lambda row: (row["국어"] + row["영어"]) / 2, axis = 1)
df
```

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "이름": ["철수", "영희", "민수", "지은", "우진", "하늘"],
    "부서": ["영업", "영업", "개발", "개발", "인사", "개발"],
    "연봉": [3500, 4000, 5000, np.nan, 3800, 5200],
    "나이": [28, 32, 27, 30, 26, 35]
}
df = pd.DataFrame(data)
print(df)
print(df.info())
print(df.describe())
```

```python
df["연봉"] = df.groupby("부서")["연봉"].transform(lambda x: x.fillna(x.mean()))
df
```

```python
def salary_grade(salary):
    if salary >= 5000:
        return "고연봉"
    elif salary >= 4000:
        return "중연봉"
    else:
        return "저연봉"
```

```python
df["연봉등급"] = df["연봉"].apply(salary_grade)
df
```

```python
summary = df.groupby("부서").agg(평균연봉=("연봉", "mean"), 인원수=("이름", "count"))
summary
```
