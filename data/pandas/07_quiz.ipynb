```python
import pandas as pd

sales_data = {
    "지점": ["강남", "강남", "홍대", "홍대", "잠실", "잠실"],
    "담당자": ["김민수", "이지은", "박철수", "최영희", "정우진", "한소희"],
    "매출": [4200, 3800, 5100, 4900, 3600, 4700],
    "고객수": [120, 95, 150, 140, 88, 130]
}

# 지점 정보 (merge 실습용, '부산'은 sales_data에 없는 지점)
branch_info = {
    "지점": ["강남", "홍대", "잠실", "부산"],
    "지역": ["서울 강남권", "서울 서북권", "서울 동남권", "부산"],
    "오픈연도": [2015, 2018, 2020, 2022]
}

# 신규 지점 실적 (concat 실습용, sales_data와 열 구성이 동일)
new_sales_data = {
    "지점": ["신촌", "신촌"],
    "담당자": ["오세훈", "윤아름"],
    "매출": [4400, 3900],
    "고객수": [110, 100]
}

sales_df = pd.DataFrame(sales_data)
branch_df = pd.DataFrame(branch_info)
new_sales_df = pd.DataFrame(new_sales_data)
```

### 그룹화 (groupby)
7. 지점별 평균 매출을 구해보세요.
8. 지점별 고객수 합계를 구해보세요.

10. 지점별로 "매출"은 평균, "고객수"는 최댓값을 각각 다르게 집계해 보세요.

```python
# 11
sales_df.groupby("지점")["매출"].mean().sort_values(ascending=False)
```

### 병합 (merge)
12. sales_df와 branch_df를 "지점" 열을 기준으로 병합하되, 양쪽에 모두 존재하는 지점만 남겨보세요. (inner join)
13. sales_df와 branch_df를 "지점" 기준으로 병합하되, 어느 한쪽에만 있는 지점(부산)까지 모두 포함하고, 값이 없는 칸이 어떻게 채워지는지 확인해 보세요. (outer join)


```python
# 12
pd.merge(sales_df, branch_df, on="지점", how="inner")
```

```python
# 13
pd.merge(sales_df, branch_df, on="지점", how="outer")
```

### 연결 (concat)
14. sales_df와 new_sales_df를 위아래로 이어 붙이고, 인덱스를 0부터 다시 매겨보세요.


```python
# 14
pd.concat([sales_df, new_sales_df], axis=0).reset_index(drop=True)
```
