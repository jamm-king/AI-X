```python
x = [10, 20, 30, 40, 50]
s = pd.Series(x)
print(s)
a = s.sum()
print("합", a)
a = s.max()
print("최대값", a)
a = s.min()
print("최소값", a)
a = s.mean()
print("평균값", a)
a = s.median()
print("중위값", a)
a = s.std()
print("표준편차", a)
```

```python
x = pd.Series([10, 20, 30, 40, 40, 40, 50, 10, 100])
y = x.value_counts().sort_values()[::-1]
print(y)
y = x[x >= 40]
print(y)
y = x[(x >= 30) & (x <= 50)]
print(y)
y = x.sort_values()
print(y)
y = x.sort_values(ascending = False)
print(y)
```

```python
data = {
    "이름": ["철수", "영희", "민수", "지은", "우진"],
    "나이": [25, 22, 28, 31, 19],
    "점수": [85, 90, 78, 92, 60]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df["이름"])
print(df[["이름"]])
print(df[["이름", "점수"]])
```

```python
print(df.loc[0, "이름"])
print(df.loc[0:2, "이름"])
print(df.iloc[0:2, 0:2])
print(df.loc[2])
print(df.iloc[-1])
```

```python
print(df["나이"] >= 25)
print(df[df["나이"] >= 25])
print(df[(df["나이"] >= 25) & (df["점수"] >= 80)])
print(df.loc[df["점수"] >= 80, ["이름", "나이"]])
```

```python
print(df[[True, False, True, True,True]])
print(df["이름"].isin(["철수", "영희"]))
print(df[df["이름"].isin(["철수", "영희"])])
print(df["나이"].between(20, 28))
print(df[df["나이"].between(20, 28)])
```
