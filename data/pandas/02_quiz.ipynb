```python
import pandas as pd
import numpy as np
```

```python
fruit_data = {
    "과일": ["사과", "바나나", "포도", "딸기", "수박"],
    "가격": [3000, 1500, 5000, 4000, 15000],
    "재고": [10, 25, 7, 15, 3]
}
```

```python
"""
1. 과일이름과 재고를 조회해 보세요.
"""

df = pd.DataFrame(fruit_data)
df[["과일", "재고"]].set_index("과일")
```

```python
"""
2. 가격이 4000원 이상인 과일만 조회해 보세요.
"""

df.loc[df["가격"] >= 4000, ["과일", "가격"]].set_index("과일")
```

```python
"""
3. 가격이 4000원 이상이면서 재고가 10개 이하인 과일을 조회해 보세요.
"""

df.loc[(df["가격"] >= 4000) & (df["재고"] <= 10), ["과일", "가격", "재고"]].set_index("과일")
```

```python
"""
4. loc을 이용해서 0번 인덱스의 "과일" 값을 조회해 보세요.
"""

df.loc[0, "과일"]
```

```python
"""
5. iloc을 이용해서 0~2번 행(포함), 0~1번 열(포함)을 조회해 보세요.
"""

df.iloc[0:3, 0:2]
```

```python
"""
6. "과일" 컬럼을 인덱스로 설정해 보세요. (원본은 바뀌지 않도록)
"""

df.set_index("과일")
```

```python
"""
7. isin()을 이용해서 "바나나"와 "수박"에 해당하는 행만 조회해 보세요.
"""

df[df["과일"].isin(["바나나", "수박"])]
```

```python
"""
8. between()을 이용해서 가격이 3000원 이상 5000원 이하인 과일을 조회해 보세요.
"""

df[df["가격"].between(3000, 5000)]
```

```python
"""
9. describe()를 이용해서 가격과 재고의 통계 요약 정보를 확인해 보세요.
"""

df[["가격", "재고"]].describe()
```
