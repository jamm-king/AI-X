```python
data = {
    "이름": ["철수", "영희", "민수"],
    "국어": [80, 90, 70],
    "영어": [85, 95, 75]
}
df = pd.DataFrame(data)
```

```python
df1 = pd.DataFrame({
    "이름": ["철수", "영희"],
    "점수": [85, 90]
})
df2 = pd.DataFrame({
    "이름": ["민수", "지은"],
    "점수": [78, 92]
})
result = pd.concat([df1, df2], axis=0, ignore_index=True)
result
```

```python
df3 = pd.DataFrame({
    "국어": [80, 90]
})
df4 = pd.DataFrame({
    "영어": [85, 95]
})
result2 = pd.concat([df3, df4], axis=1)
result2
```

```python
students = pd.DataFrame({
    "학번": [1, 2, 3],
    "이름": ["철수", "영희", "민수"]
})
scores = pd.DataFrame({
    "학번": [1, 2, 4],
    "점수": [85, 90, 70]
})
result3 = pd.merge(students, scores, on="학번")
result3
```

```python
result4 = pd.merge(students, scores, on="학번", how="left")
result4
```

```python
pd.merge(students, scores, on="학번", how="right")
```

```python
result5 = pd.merge(students, scores, on="학번", how="outer")
result5
```
